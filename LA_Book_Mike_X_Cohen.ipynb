```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the vector
v = np.array([2, -1])

# 2. Setup the plot background and grids
plt.figure(figsize=(5, 5)) # Makes the plot perfectly square
plt.grid(True, which='both', linestyle='--', linewidth=0.5, color='gray')

# Draw major X and Y axis lines through the origin (0,0)
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

# 3. Plot the vector with an arrowhead
# plt.quiver parameters: X_start, Y_start, X_direction, Y_direction
plt.quiver(1, 2, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='royalblue', label='Vector v')

# 4. Set boundaries and display labels
plt.axis([-3, 3, -3, 3])
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('2D Vector Plot')
plt.legend()

# 5. Render the plot
plt.show()
```
